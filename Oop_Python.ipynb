```python
class Student:
  def __init__(self, name, marks):
    self.name = name
    self.marks = marks

  def get_avg(self):
    total_sum = 0
    for val in self.marks:
      total_sum = total_sum + val
    print("Student Name: ", self.name, "|| Average marks: ", total_sum/len(self.marks))

s1 = Student("Mujahid", [90,95,94])
print(s1.name)
s1.get_avg()
```

```python
# Abstraction
# hiding the implementation details of a class and
# only showing the essential features to the user

class Car:
  def __init__(self):
    self.acc = False
    self.brk = False
    self.clutch = False

  def start(self):
    self.acc = True
    self.clutch = True
    print ("Car Started")


car1 = Car()
car1.start()
```

```python
class Account:
  def __init__(self, bal, acc):
    self.balance = bal
    self.account_number = acc

  def debit(self, amount):
    self.balance -= amount
    print ("Amount has been debited: ", amount)
    print ("Remaining Balance: ", self.balance)

  def deposit(self, amount):
    self.balance += amount
    print ("Amount has been deposited: ", amount)
    print ("Remaining Balance: ", self.balance)

acc1 = Account(1000, 123456789)
print(acc1.balance,acc1.account_number)
acc1.debit(200)
acc1.deposit(2000)
```

```python
## Super Class (jb child class s parent class k constructor method ko use kren)

class Car:
  def __init__(self, type):
    self.type = type

  @staticmethod
  def start():
    print ("Car Started")

  @staticmethod
  def stop():
    print ("Car Stopped")


class ToyotaCar(Car):
  def __init__(self, name, type):
    super().__init__(type)
    self.name = name


car1 = ToyotaCar("Prius", "Hybrid")
print(car1.name, car1.type)
```

```python
## Class Method
# jsmn first parameter self ki bjaye obj hota hai

class Person:
  name = 'anonymous'

  @classmethod
  def changeName (cls, name):
    cls.name = name   ## yh b class k attribute ko chng krega


p1 = Person()
p1.changeName("Mujahid")
print(p1.name)

print(Person.name)
```

```python
#Property
# WE use @property decorator on any method in the class to use the method as a property.

class Student:
  def __init__(self, phy,chem,math):
    self.phy = phy
    self.chem = chem
    self.math = math

  # def calcPercentage(self):
  #   self.percentage = str((self.phy+self.chem+self.math) / 3) + "%"
  #rather than above code we can do the below one

  @property
  def calcPercentage(self):
    return str((self.phy+self.chem+self.math) / 3) + "%"

stu1 = Student(90,95,94)
print(stu1.calcPercentage)

stu1.phy = 69
print(stu1.calcPercentage)

```

```python
class Complex:
  def __init__(self, real, img):
    self.real = real
    self.img = img

  def showNumber(self):
    print(self.real, "i +", self.img, "j")

  ## Dunder Function
  def __add__(self, other):
    print ("------------")
    return Complex(self.real + other.real, self.img + other.img)

  def __sub__(self, other):
    print ("------------")
    return Complex(self.real - other.real, self.img - other.img)

num1 = Complex(2,3)
num1.showNumber()

num2 = Complex(3,4)
num2.showNumber()

num3 = num1 + num2
num3.showNumber()

print()
print()
print()

num1 = Complex(2,3)
num1.showNumber()

num2 = Complex(3,4)
num2.showNumber()

num4 = num1 - num2
num4.showNumber()
```

```python
class Circle:
  def __init__(self, radius):
    self.radius = radius

  def area(self):
    return 3.14 * self.radius * self.radius

  def perimeter(self):
    return 2 * 3.14 * self.radius

c1 = Circle(5)
print ("Radius: ", c1.radius)
print("Area: " , c1.area())
print("Perimeter: ", c1.perimeter())
```

```python
class Employee:
  def __init__(self, role, dept, salary):
    self.role = role
    self.dept = dept
    self.salary = salary

  def showDetails(self):
    print("Role: ", self.role)
    print("Department: ", self.dept)
    print("Salary: ", self.salary)

class Engineer (Employee):
  def __init__(self, name, age):
    self.name = name
    self.age = age
    super().__init__("Engineer", "IT", "70000")


emp1 = Engineer("Mujahid", 22)
emp1.showDetails()
```

```python
class Order:
  def __init__(self, item, price):
    self.item = item
    self.price = price

  #true if self.price is greater than odr2.price else false
  def __gt__(self, odr2):
    return self.price > odr2.price


ord1 = Order("chips", 40)
ord2 = Order("burger", 35)

print(ord1 > ord2)
```
